```python
import matplotlib.pyplot as plt
import numpy as np

# Data
pollutants = ["pm25_median", "pm10_median", "o3_median", "no2_median", "so2_median", "co_median"]

# RMSE and MAPE values for each method
rmse_without = [17.013, 10.502, 7.134, 4.171, 0.691, 1.236]
rmse_mean = [13.500, 6.594, 7.813, 3.590, 1.288, 4.313]
rmse_median = [13.511, 6.625, 7.856, 3.543, 1.284, 4.276]
rmse_gcn = [14.3502, 9.0393, 10.1815, 4.4590, 1.2830, 5.1738]
rmse_csf = [14.5832, 6.249076586836916, 8.368310259287103, 3.744805948227834, 1.1131687961310404, 4.65304155012427]



# Bar width
bar_width = 0.25
x = np.arange(len(pollutants))

# Plot RMSE
fig, ax = plt.subplots(figsize=(8, 6))

ax.bar(x - bar_width, rmse_without, bar_width, label="Without Imputation", color="skyblue")
ax.bar(x, rmse_mean, bar_width, label="Mean Imputation", color="orange")
ax.bar(x + bar_width, rmse_median, bar_width, label="Median Imputation", color="green")
ax.set_title("RMSE Comparison")
ax.set_xticks(x)
ax.set_xticklabels(pollutants, rotation=45)
ax.set_ylabel("RMSE")
ax.legend()


# Adjust layout
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 6))

ax.bar(x - bar_width, rmse_mean, bar_width, label="SVR", color="orange")
ax.bar(x, rmse_gcn, bar_width, label="GCN", color="blue")
ax.bar(x + bar_width, rmse_csf, bar_width, label="CSF Model", color="green")

# Customize plot
ax.set_title("RMSE Comparison Across Methods (Mean Imputation)")
ax.set_xticks(x)
ax.set_xticklabels(pollutants, rotation=45)
ax.set_ylabel("RMSE")
ax.legend()

# Adjust layout and show plot
plt.tight_layout()
plt.show()
```
