```python
import matplotlib.pyplot as plt
import numpy as np

# Data
pollutants = ["pm25_median", "pm10_median", "o3_median", "no2_median", "so2_median", "co_median"]

# RMSE and MAPE values for each method
#svr_rmse_without = [17.013, 10.502, 7.134, 4.171, 0.691, 1.236]
#svr_rmse_mean = [13.500, 6.594, 7.813, 3.590, 1.288, 4.313]
#svr_rmse_median = [13.511, 6.625, 7.856, 3.543, 1.284, 4.276]
rmse_gcn_mean = [14.1090, 7.7614, 8.5564, 4.1187, 1.1970, 5.0526]
#rmse_gcn_mean = [14.0116, 7.6102, 9.2456, 4.1582, 1.1957, 5.4959]
rmse_gcn_median = [14.0579, 7.4575, 9.3622,4.1378, 1.1669, 5.4737]
rmse_gcn_without = [17.3972, 9.1230, 7.0230,6.3222 ,0.9335, 1.6747]

rmse_csf_mean = [14.969435889174594, 6.222229567635728, 8.231036366838369, 3.802360209399013, 
                 1.1279438293592627, 5.0577452124201745]

rmse_svr_without = [14.687, 8.572, 5.530, 3.506, 0.271, 0.979]
rmse_svr_mean = [13.707, 6.183, 6.531, 3.332, 1.155, 3.826]
rmse_svr_median = [13.768, 6.241, 6.552, 3.349, 1.186, 3.883]

#rmse_fda_mean = [13.4258, 7.7084, 6.5067, 3.6398, 1.1074, 4.0836]
rmse_fda_mean = [13.3131, 7.6096, 6.6928, 3.6537, 1.1081, 4.4822]

rmse_csf_mean_extreme = [21.8909, 10.5522, 10.1826, 6.4974, 2.4693, 10.1584]
rmse_gcn_mean_extreme = [17.7159, 10.5692, 8.9026, 4.5606, 2.1427, 5.4692]
rmse_fda_mean_extreme = [20.3153, 12.9575, 8.4232, 6.4489, 2.4741, 9.3938]
rmse_svr_mean_extreme = [21.768, 15.176, 8.064, 6.884, 1.961, 9.412]


# Bar width
bar_width = 0.25
x = np.arange(len(pollutants))

# Plot RMSE
fig, ax = plt.subplots(figsize=(8, 6))

ax.bar(x - bar_width, rmse_svr_without, bar_width, label="Without Imputation", color="skyblue")
ax.bar(x, rmse_svr_mean, bar_width, label="Mean Imputation", color="orange")
ax.bar(x + bar_width, rmse_svr_median, bar_width, label="Median Imputation", color="green")
ax.set_title("SVR RMSE Comparison")
ax.set_xticks(x)
ax.set_xticklabels(pollutants, rotation=45)
ax.set_ylabel("RMSE")
ax.legend()


# Adjust layout
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

for i, pollutant in enumerate(pollutants):
    ax = axes[i]
    # Plot lines for each method
    ax.plot([1, 2], [rmse_gcn_mean[i], rmse_gcn_mean_extreme[i]], label="GCN", marker='o')
    ax.plot([1, 2], [rmse_csf_mean[i], rmse_csf_mean_extreme[i]], label="CosSquareFormer", marker='o')
    ax.plot([1, 2], [rmse_svr_mean[i], rmse_svr_mean_extreme[i]], label="SVR", marker='o')
    ax.plot([1, 2], [rmse_fda_mean[i], rmse_fda_mean_extreme[i]], label="FDA", marker='o')
    
    # Customize plot
    ax.set_title(f"RMSE Comparison for {pollutant}")
    ax.set_xticks([1, 2])
    ax.set_xticklabels(["Normal", "Extreme"])
    ax.set_ylabel("RMSE")
    ax.set_xlabel("Condition")
    ax.legend()
    ax.grid(True)

# Adjust layout and show all plots
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

for i, pollutant in enumerate(pollutants):
    ax = axes[i]
    
    # Scatter points for each model
    ax.scatter(rmse_gcn_mean[i], rmse_gcn_mean_extreme[i], label="GCN", color="blue", s=100)
    ax.scatter(rmse_csf_mean[i], rmse_csf_mean_extreme[i], label="CSF", color="green", s=100)
    ax.scatter(rmse_svr_mean[i], rmse_svr_mean_extreme[i], label="SVR", color="orange", s=100)
    ax.scatter(rmse_fda_mean[i], rmse_fda_mean_extreme[i], label="FDA", color="purple", s=100)
    
    # Add labels for each point
    ax.text(rmse_gcn_mean[i], rmse_gcn_mean_extreme[i], "GCN", fontsize=10, color="blue")
    ax.text(rmse_csf_mean[i], rmse_csf_mean_extreme[i], "CSF", fontsize=10, color="green")
    ax.text(rmse_svr_mean[i], rmse_svr_mean_extreme[i], "SVR", fontsize=10, color="orange")
    ax.text(rmse_fda_mean[i], rmse_fda_mean_extreme[i], "FDA", fontsize=10, color="purple")
    
    # Plot y=x line
    max_val = max(rmse_gcn_mean_extreme[i], rmse_csf_mean_extreme[i], rmse_svr_mean_extreme[i], rmse_fda_mean_extreme[i]) + 1
    ax.plot([0, max_val], [0, max_val], linestyle="--", color="gray", label="Normal=Extreme")
    
    # Customize plot
    ax.set_title(f"{pollutant}")
    ax.set_xlabel("Normal RMSE")
    ax.set_ylabel("Extreme RMSE")
    ax.legend()
    ax.grid(True)

# Adjust layout and show all plots
plt.tight_layout()
plt.show()
```
